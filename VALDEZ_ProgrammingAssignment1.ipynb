### **A. WORD ROTATION PROBLEM**

```python
def rotate_word(text): #defines name and function of variable
#slices the characters from index 1 to the end then takes the character from index 0 and places it at the end
  return text[1:] + text[0]
```

```python
print(rotate_word("python"))
print(rotate_word("logic"))
print(rotate_word("Code"))
print(rotate_word("A"))
```

### **B. USERNAME BUILDER PROBLEM**

```python
def make_username(first_name, last_name): #defines name and function of variables
  #.lower() converts all letters to lowercase then .replace(" ","") removes the spaces
  new_first = first_name.lower().replace(" ","")
  new_last = last_name.lower().replace(" ","")
  #connects the new first and last name with a period in between
  complete_username = new_first + "." + new_last
  return complete_username
```

```python
print(make_username("Ada", "Lovelace"))
print(make_username("Alan", "Turing"))
print(make_username("Ana Maria", "De Leon"))
```

### **C. BOOKEND SWAP PROBLEM**

```python
def swap_bookends(items): #defines name and function of variables
#uses *middle to store all elements between the first and last index to a list
  first, *middle, last = items
  #returns list with first and last index swapped while the middle order is unchanged
  return [last] + middle + [first]
```

```python
print(swap_bookends([1,2,3,4,5,6]))
print(swap_bookends(["red","green","blue"]))
print(swap_bookends([8,3]))
```
